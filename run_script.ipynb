```python
import matplotlib.pyplot as plt
import numpy as np
from skimage import io
import os
```

```python
path = 'predictions'
idx = 'last'
frame = 0
output = io.imread(os.path.join(path, f'output_{idx}_pix.tif'))
trace = io.imread(os.path.join(path, f'trace_pix.tif'))
retrace = io.imread(os.path.join(path, f'retrace_pix.tif'))
f,ax = plt.subplots(1,3)
ax[0].imshow(output[frame], cmap='gray', interpolation='None')
ax[1].imshow(trace[frame], cmap='gray', interpolation='None')
ax[2].imshow(retrace[frame], cmap='gray', interpolation='None')
```

```python
ensemble = '0'
t2r_loss = np.genfromtxt(f"checkpoints/ensemble{ensemble}/t2r_loss.csv")
r2t_loss = np.genfromtxt(f"checkpoints/ensemble{ensemble}/r2t_loss.csv")
plt.plot(t2r_loss, label='t2r')
plt.plot(r2t_loss, label='r2t')
plt.legend()
plt.xlabel('iterations')
plt.ylabel('loss')
# plt.yscale('log')
```

```python
ensemble = '0'
t2r_lr = np.genfromtxt(f"checkpoints/ensemble{ensemble}/t2r_lr.csv")
r2t_lr = np.genfromtxt(f"checkpoints/ensemble{ensemble}/r2t_lr.csv")
plt.plot(t2r_lr, label='t2r')
plt.plot(r2t_lr, label='r2t')
plt.legend()
plt.xlabel('iterations')
plt.ylabel('learning rate')
```
